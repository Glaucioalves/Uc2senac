```python
url = 'https://www.ispdados.rj.gov.br/Arquivos/BaseDPEvolucaoMensalCisp.csv'
df = pd.read_csv(url, encoding='latin-1', sep=';')
df
```

```python
# agrupamento por cisp roubo veiclo 
df_cisp = df.groupby('cisp')['roubo_veiculo'].sum().reset_index()

# ordenado top 15 + e top - 5
df_cisp = df_cisp.sort_values(by='roubo_veiculo')
df_cisp
```

```python
df_capital = df.loc[df['munic'] == 'Rio de Janeiro']

# Usando .loc
df_capital = df.loc[df['munin'] == 'rio de janeiro']

# Usando .query() — sintaxe alternativa mais legível
df_capital = df.query("munic == 'Rio de Janeiro'")
```

```python
df_cisp_furto = df_capital.groupby('cisp')['furto_veiculos'].sum().reset_index
df_cisp_furto = df_cisp_furto.sort_values(by='furto_veiculos',ascending=False )
df_cisp_furto.head(10)
```

```python
array_cisp_capital_furto = np.array(df_cisp_furto['furto_veiculo'])
array_cisp_capital_furto
```
